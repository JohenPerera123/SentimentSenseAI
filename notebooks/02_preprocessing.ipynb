# SentimentSense AI - Preprocessing Demonstration

This notebook demonstrates the preprocessing pipeline on IMDb and Twitter samples.

```python
import pandas as pd
from src.preprocessing.preprocessing_pipeline import PreprocessingPipeline
from src.preprocessing.text_cleaner import clean_text_basic

imdb_pipeline = PreprocessingPipeline(is_twitter=False)
twitter_pipeline = PreprocessingPipeline(is_twitter=True)
```

## 1. IMDb Examples (HTML, Negation)

```python
imdb_examples = [
    "This movie is <b>great</b><br /> but I did not like the ending.",
    "I have never seen such a bad film!!!",
    "Absolutely no plot, 0/10 would not recommend."
]

print(f"{'Original Text':<55} | {'Processed Text'}")
print("-" * 90)
for text in imdb_examples:
    print(f"{text[:50]:<55} | {imdb_pipeline.process(text)}")
```

## 2. Twitter Examples (Mentions, URLs, Hashtags)

```python
twitter_examples = [
    "@user1 check this out https://example.com #AmazingMovie",
    "I loooove this so much!!!!",
    "I can't believe it wasn't better than the prequel."
]

print(f"{'Original Text':<55} | {'Processed Text'}")
print("-" * 90)
for text in twitter_examples:
    print(f"{text[:50]:<55} | {twitter_pipeline.process(text)}")
```
