## Inference on Pre-trained Models
Using the `DistilBertPredictor` trained in Phase 4.

```python
from src.models.distilbert_predictor import DistilBertPredictor

# If you have trained the model via train_distilbert.py
try:
    predictor = DistilBertPredictor('../models/distilbert/imdb')
    texts = [
        "This movie was absolutely wonderful! I highly recommend it.",
        "What a waste of time. The acting was terrible and the plot made no sense."
    ]
    
    for t in texts:
        res = predictor.predict(t)
        sentiment = 'Positive' if res['sentiment'] == 1 else 'Negative'
        print(f"\nText: {t}\nSentiment: {sentiment}\nConfidence: {res['confidence']:.4f}")
except Exception as e:
    print("Model not found. Run scripts/train_distilbert.py first.")
```
